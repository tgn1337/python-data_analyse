```python
import numpy
import scipy.sparse
import time

#4.1
arr = numpy.random.randint(0, 10, size=(5, 5)).astype(float)

mask = numpy.random.rand(5, 5) > 0.2
arr[mask] = 0

print(arr)

arr_csr = scipy.sparse.csr_matrix(arr)
print("\n", arr_csr)

#4.2
arr_coo = arr_csr.tocoo()
arr_csc = arr_csr.tocsc()

print("\n", arr_coo)
print("\n", arr_csc)

#4.3
vec = [1, 1, 1, 1, 1]
result = arr_csr @ vec
print("\n", result)

#4.4
BigArr = numpy.zeros((1000, 1000))

indices = numpy.random.choice(1000 * 1000, 10000, replace=False)
BigArr.flat[indices] = numpy.random.rand(10000)

BigSparse = scipy.sparse.csr_matrix(BigArr)

vec2 = numpy.random.rand(1000)
%timeit BigArr @ vec2
%timeit BigSparse @ vec2
```
