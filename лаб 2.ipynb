```python
#подключение
import pandas as pd
import numpy as np
import data
```

```python
# 1.1
my_series = pd.Series([100, 200, 300, 400], index=['a', 'b', 'c', 'd'])
print("1.1:")
print(my_series, "\n")

#1.2
df = pd.DataFrame(data.data)
display(df)

#1.3
print("\n1.3 Базовый осмотр основного df:")
print("Первые 3 строки:")
display(df.head(3))

print("Последние 3 строк:")
display(df.tail(3))

print("Информация о DataFrame:")
df.info()

print("\nСтатистика:")
display(df.describe())
```

```python
#2.1
it_employees = df[df['department'] == 'IT'][['name', 'salary']]
print("2.1:")
display(it_employees)

#2.2
print("\n2.2:")
display(df.iloc[:3, :2])

#2.3
print("\n2.3:")
display(df[df['salary'] > 70000])

#2.4
df['total_comp'] = df['salary'] + df['bonus'].fillna(0)
print("\n2.4:")
display(df[['name', 'salary', 'bonus', 'total_comp']])

#2.5
df = df.drop(columns=['hire_date'])
print("\n2.5:")
display(df)
```

```python
#3.1
print("3.1:")
print(df.isna().sum())

#3.2
df['salary'] = df.groupby('department')['salary'].transform(lambda x: x.fillna(x.mean()))
print("\n3.2:")
display(df[['name', 'department', 'salary']])

#3.3
df['bonus'] = df['bonus'].fillna(0)

#3.4
df = df.dropna(subset=['salary'])
print("\n3.4:")
display(df)
```

```python
#4.1
mean_salary = df.groupby('department')['salary'].mean()
print("4.1:")
print(mean_salary)

#4.2
dept_stats = df.groupby('department').agg(
    mean_salary=('salary', 'mean'),
    max_bonus=('bonus', 'max'),
    emp_count=('name', 'count')
)
print("\n4.2:")
display(dept_stats)

#4.3
pivot = df.pivot_table(index='department', values='salary', aggfunc='mean')
print("\n4.3:")
display(pivot)

#4.4
print("\n4.4:")
print(mean_salary.sort_values(ascending=False))
```

```python
#5.1
managers = pd.DataFrame({
    'department': ['IT', 'HR', 'Finance'],
    'manager': ['Радченко', 'Радченко2', 'Радченко3']
})
print("5.1")
display(managers)

#5.2
df_merged = df.merge(managers, on='department', how='left')
print("\n5.2:")
display(df_merged)

#5.3
new_employee = pd.DataFrame([{
    'name': 'Никита',
    'department': 'IT',
    'salary': 95000,
    'bonus': 7000,
    'total_comp': 102000,
    'manager': 'Иванов'
}])
df_concat = pd.concat([df_merged, new_employee], ignore_index=True)
print("\n5.3:")
display(df_concat.tail())

#5.4
df_final = df_concat.reset_index(drop=True)
print("\n5.4:")
print(df_final.index.tolist())
```

```python
#6.1
top3 = df_final.nlargest(3, 'total_comp')[['name', 'department', 'total_comp']]
print("6.1:")
display(top3)

#6.2
dept_avg = df_final.groupby('department')[['salary', 'bonus']].mean()
print("\n6.2:")
display(dept_avg)

#6.3
high_salary_share = df_final.groupby('department')['salary'].apply(lambda x: (x > 75000).mean())
print("\n6.3:")
print(high_salary_share)
```
